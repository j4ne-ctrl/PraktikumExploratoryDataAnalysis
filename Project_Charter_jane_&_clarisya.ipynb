# 1. Import Library


```python
# 1. Import Library
import pandas as pd
import numpy as np
```

# 2. Data Loading & Inspection


```python
# 2. Data Loading & Inspection
df = pd.read_csv  ("dataset_nilai_akademik_siswa (1).csv")

print(df.head())
print()
print(df.info())
print()
print(df.describe())
print()
print(df.shape)
```

# 3. Data Cleaning


```python
# 3. Data Cleaning

# menghapus data duplikat
df = df.drop_duplicates()

# menangani missing value
df['nilai'] = df['nilai'].astype(str).str.replace(',', '.')
df['nilai'] = pd.to_numeric(df['nilai'], errors='coerce')

# Mengisi missing value pada 'nilai' dengan 0, dan 'guru_pengampu' dengan 'Belum Diisi'
df['nilai'] = df['nilai'].fillna(0)
df['guru_pengampu'] = df['guru_pengampu'].fillna('Belum Diisi')
```

# 4. Data Manipulation


```python
# 4. Data Manipulation

# a. Filtering: Siswa yang mendapat nilai di bawah KKM (< 75)
remidi = df[df['nilai'] < 75]
print("\n--- Data Siswa Remidi (Nilai < 75) ---")
print(remidi[['nama', 'kelas', 'mata_pelajaran', 'nilai']].head())

# b. Sorting: Mengurutkan dari nilai terkecil ke tertinggi
urut_nilai = df.sort_values(by='nilai', ascending=True)

# c. Kolom Turunan: Menambahkan kolom 'status' (Lulus / Remidi)
df['status'] = np.where(df['nilai'] >= 75, 'Lulus', 'Remidi')

# d. Groupby / Agregasi: Rata-rata nilai per mata pelajaran
rata_matpel = df.groupby('mata_pelajaran')['nilai'].mean()
print("\n--- Rata-rata Nilai Per Mata Pelajaran ---")
print(rata_matpel)
```

# 5. Menyimpan Dataset Cleaned & Ringkasan Temuan


```python
# 5. Menyimpan Dataset Cleaned & Ringkasan Temuan

# Menyimpan hasil dataset yang sudah bersih ke file CSV baru
df.to_csv('dataset_nilai_akademik_clean.csv', index=False)

# Ringkasan Temuan Awal (Profiling Summary)
print("\n--- RINGKASAN TEMUAN AWAL ---")
print(f"1. Total siswa yang perlu meremidi (nilai < 75) sebanyak {len(remidi)} catatan transaksi.")
print(f"2. Mata pelajaran dengan rata-rata nilai terendah adalah: {rata_matpel.idxmin()} ({rata_matpel.min():.2f}).")
print("3. Data bersih berhasil disimpan sebagai 'dataset_nilai_akademik_clean.csv'.")
```
